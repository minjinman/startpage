데이터를 넣으면 계획 → 실행 → 한국어 HTML 리포트까지 만듭니다. (먼저 `pip install -e .` 실행)

```python
import pandas as pd, numpy as np
from autoeda import analyze, plan_analysis
```

## 1. 샘플 데이터 만들기 (실제로는 `analyze("내파일.csv")`)

```python
rng = np.random.default_rng(0)
n = 300
line = rng.choice(['1라인','2라인','3라인'], n)
df = pd.DataFrame({
    '라인': line,
    '온도': 70 + rng.normal(0, 2, n) + (line == '3라인') * 3,
    '압력': rng.normal(5, .4, n),
    '생산량': rng.integers(900, 1500, n),
})
df['불량수'] = (df['온도'] - 68).clip(0) * rng.random(n)
df.head()
```

## 2. 계획 먼저 보기 (가정과 선택 이유가 나옵니다)

```python
plan = plan_analysis(df, group='라인')
plan
```

## 3. 계획을 고치고 실행

```python
plan.drop('outliers')   # 필요 없는 단계 제외
report = plan.run()
report.show()
```

## 4. 파일로 저장

```python
report.to_html('report.html')
report.to_brief('llm_brief.md', anonymize=True)   # Claude 채팅에 붙여넣어 해석받기용 요약
```
